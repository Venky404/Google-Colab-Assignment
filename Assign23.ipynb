Setup, Load Dataset & Preprocessing

```python
# Import required libraries

import pandas as pd
import numpy as np

from google.colab import files

from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import OneHotEncoder, StandardScaler

from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
uploaded = files.upload()
```

```python
df = pd.read_csv("Most Streamed Artists on Spotify (17_07_2026) V1.1.csv")

df.head()
```

```python
print("Dataset Shape :", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
median_streams = df["Total Streams (in millions)"].median()

df["Popularity"] = np.where(
    df["Total Streams (in millions)"] >= median_streams,
    1,
    0
)

df[["Total Streams (in millions)", "Popularity"]].head()
```

```python
# Regression Target (Q1)

X_reg = df.drop(columns=["Total Streams (in millions)", "Popularity"])

y_reg = df["Total Streams (in millions)"]

# Classification Target (Q2,Q3,Q4,Q5)

X_cls = df.drop(columns=["Total Streams (in millions)", "Popularity"])

y_cls = df["Popularity"]
```

```python
numerical_columns = X_cls.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_columns = X_cls.select_dtypes(
    include=["object"]
).columns
```

```python
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numerical_columns),
    ("cat", categorical_transformer, categorical_columns)
])
```

```python
# Linear Regression

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.20,
    random_state=42
)

# Classification Models

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls,
    y_cls,
    test_size=0.20,
    random_state=42
)
```

```python
# Linear Regression

X_train_reg = preprocessor.fit_transform(X_train_reg)
X_test_reg = preprocessor.transform(X_test_reg)

# Classification Models

X_train_cls = preprocessor.fit_transform(X_train_cls)
X_test_cls = preprocessor.transform(X_test_cls)

# Dense arrays for Naive Bayes

X_train_cls_dense = X_train_cls.toarray()
X_test_cls_dense = X_test_cls.toarray()

print("Preprocessing completed successfully.")
```

Q1. Linear Regression

```python
# Train Linear Regression Model

linear_model = LinearRegression()

linear_model.fit(X_train_reg, y_train_reg)
```

```python
# Predict the test data

y_pred_reg = linear_model.predict(X_test_reg)
```

```python
# Calculate R² Score

r2 = r2_score(y_test_reg, y_pred_reg)

print("R² Score :", round(r2, 4))
```

```python
# Compare Actual and Predicted Values

prediction_df = pd.DataFrame({
    "Actual Streams": y_test_reg.values,
    "Predicted Streams": y_pred_reg
})

prediction_df.head(10)
```

```python
print("Q1 Completed Successfully")
print("The Linear Regression model was trained.")
print("The model was evaluated using the R² Score.")
```

Q2. Logistic Regression

```python
# Train Logistic Regression Model

logistic_model = LogisticRegression(max_iter=1000)

logistic_model.fit(X_train_cls, y_train_cls)
```

```python
# Predict the test data

y_pred_log = logistic_model.predict(X_test_cls)
```

```python
cm = confusion_matrix(y_test_cls, y_pred_log)

print("Confusion Matrix")
print(cm)
```

```python
accuracy = accuracy_score(y_test_cls, y_pred_log)

precision = precision_score(y_test_cls, y_pred_log)

recall = recall_score(y_test_cls, y_pred_log)

f1 = f1_score(y_test_cls, y_pred_log)

print("Accuracy :", round(accuracy,4))
print("Precision:", round(precision,4))
print("Recall   :", round(recall,4))
print("F1 Score :", round(f1,4))
```

```python
print(classification_report(y_test_cls, y_pred_log))
```

```python
print("Q2 Completed Successfully")
print("The Logistic Regression model was trained.")
print("The model was evaluated using:")
print("- Confusion Matrix")
print("- Accuracy")
print("- Precision")
print("- Recall")
print("- F1 Score")
```

Q3. K-Nearest Neighbors (KNN)

```python
# Train KNN models with different K values

k_values = [3, 5, 7]
accuracy_list = []

for k in k_values:
    knn_model = KNeighborsClassifier(n_neighbors=k)

    knn_model.fit(X_train_cls, y_train_cls)

    y_pred_knn = knn_model.predict(X_test_cls)

    accuracy = accuracy_score(y_test_cls, y_pred_knn)

    accuracy_list.append(accuracy)

    print(f"K = {k} --> Accuracy = {accuracy:.4f}")
```

```python
# Compare accuracy of different K values

knn_results = pd.DataFrame({
    "K Value": k_values,
    "Accuracy": accuracy_list
})

knn_results
```

```python
best_accuracy = max(accuracy_list)
best_k = k_values[accuracy_list.index(best_accuracy)]

print("Best K Value :", best_k)
print("Best Accuracy :", round(best_accuracy, 4))
```

```python
print("Q3 Completed Successfully")
print("The KNN model was trained with K = 3, 5, and 7.")
print("The model with the highest accuracy was selected as the best.")
```

Q4. Naive Bayes

```python
# Train Gaussian Naive Bayes Model

naive_model = GaussianNB()

naive_model.fit(X_train_cls_dense, y_train_cls)
```

```python
# Predict test data

y_pred_nb = naive_model.predict(X_test_cls_dense)
```

```python
cm_nb = confusion_matrix(y_test_cls, y_pred_nb)

print("Confusion Matrix")
print(cm_nb)
```

```python
print(classification_report(y_test_cls, y_pred_nb))
```

```python
accuracy_nb = accuracy_score(y_test_cls, y_pred_nb)

print("Accuracy :", round(accuracy_nb, 4))
```

```python
print("Q4 Completed Successfully")
print("The Gaussian Naive Bayes model was trained.")
print("The model was evaluated using the Confusion Matrix,")
print("Classification Report, and Accuracy.")
```
