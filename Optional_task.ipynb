```python
import pandas as pd

df = pd.read_csv("marvel_master.csv")

df.head(10)
```

```python
print("Shape of Dataset:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
df.describe()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
numerical = df.select_dtypes(include=['int64', 'float64']).columns

categorical = df.select_dtypes(include=['object', 'bool']).columns

print("Numerical Columns:")
print(numerical)

print("\nCategorical Columns:")
print(categorical)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

columns = ['year', 'runtime_min', 'imdb_rating', 'popularity']

fig, axes = plt.subplots(2, 2,)

for ax, col in zip(axes.flat, columns):
    sns.histplot(df[col], kde=True, ax=ax, color='skyblue')
    ax.set_title(f'Distribution of {col}')
    ax.set_xlabel(col)
    ax.set_ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
cat_cols = ['type', 'rated', 'universe']

fig, axes = plt.subplots(1,3, figsize=(18,5))

for ax, col in zip(axes, cat_cols):
    sns.countplot(data=df, x=col, ax=ax)
    ax.set_title(f'Count Plot of {col}')
    ax.tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,8))

corr = df.select_dtypes(include=['int64','float64']).corr()

sns.heatmap(corr,
            annot=True,
            cmap='coolwarm',
            fmt=".2f")

plt.title("Correlation Heatmap")

plt.show()
```

```python
print("Average IMDb Rating:")
print(df['imdb_rating'].mean())

print("\nMaximum IMDb Rating:")
print(df['imdb_rating'].max())

print("\nMinimum IMDb Rating:")
print(df['imdb_rating'].min())

print("\nAverage IMDb Rating by Type:")
print(df.groupby('type')['imdb_rating'].mean())
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(data=df,
            x='type',
            y='imdb_rating')

plt.title("IMDb Rating by Type")

plt.show()
```

```python
print("1. Average Release Year:")
print(df['year'].mean())

print("\n2. Average IMDb Rating:")
print(df['imdb_rating'].mean())

print("\n3. Average IMDb Rating by Type (Movie/Series):")
print(df.groupby('type')['imdb_rating'].mean())

print("\n4. Number of Titles in Each Universe:")
print(df['universe'].value_counts())

print("\n5. Most Common Universe:")
print(df['universe'].mode()[0])
```
